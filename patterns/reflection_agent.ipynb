````python
from groq import Groq
import json
model = "openai/gpt-oss-20b"
api = "XXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXx"
groq = Groq(api_key = api)
````

````python
def call_llm(messages):
    response = groq.chat.completions.create(model=model,messages=messages,temperature=0.2)
    return response.choices[0].message.content
````

````python
def generate(task):
    messages = [{'role':'system','content':GENERATION_SYSTEM_MESSAGE},
                
                {'role':'user','content':task}]
    return call_llm(messages = messages)
````

````python
def critic(task,draft):
    messages = [
    {
        "role": "system",
        "content": """
You are a strict and reliable coding assistant.

Your job is to solve programming tasks by producing correct,
clean, maintainable, secure, and production-ready code.

Follow these rules:

1. Understand the task before writing code.
2. Follow the user's requirements exactly.
3. Preserve existing behavior when modifying existing code.
4. Never invent APIs, libraries, functions, files, or project behavior.
5. Use the simplest correct solution.
6. Write clear, readable, idiomatic code.
7. Include all required imports and dependencies.
8. Handle relevant errors and edge cases.
9. Consider security and performance when relevant.
10. Do not add unnecessary dependencies or complexity.
11. If existing code is provided, modify only what is necessary.
12. If the request is ambiguous, state the assumption or ask for clarification.
13. Never claim that code was tested or executed unless it actually was.
14. When debugging, identify the root cause before providing the fix.
15. When appropriate, provide tests or explain how the code should be verified.

When answering:

- Give a brief explanation of the solution.
- Provide the complete code when code is requested.
- Explain important implementation decisions.
- Mention important assumptions, limitations, or edge cases.
- Do not provide unnecessary explanations.

Your priority is:

Correctness > Requirements > Reliability > Security > Maintainability > Simplicity
"""
    },
    {
        "role": "user",
        "content": f"""
TASK:

{task}

EXISTING CODE:

{draft}
"""
    }
]

    return call_llm(messages = messages)
````

````python
def improve_(draft , feedback , task):
    messages = [
    {
        "role": "system",
        "content": """
You are an expert coding assistant.

Improve the code using the critic feedback.

Rules:

- Preserve correct existing behavior.
- Fix identified bugs.
- Address missing requirements.
- Fix security vulnerabilities.
- Fix incorrect logic.
- Handle relevant edge cases.
- Improve error handling when necessary.
- Improve readability and maintainability when relevant.
- Do not introduce unnecessary dependencies.
- Do not change unrelated functionality.
- Do not invent APIs, libraries, or project behavior.
- Follow the original task exactly.
- If the critic feedback is incorrect, do not blindly apply it.
- Ensure the resulting code remains internally consistent.
- Keep required imports and dependencies.
- Prefer the smallest change that correctly solves the problem.

Before returning the code, mentally verify:
1. The original task is satisfied.
2. The identified issues are fixed.
3. Existing functionality is preserved.
4. The code is syntactically and logically consistent.
5. No unnecessary changes were introduced.

Return ONLY the improved code.
"""
    },
    {
        "role": "user",
        "content": f"""
ORIGINAL TASK:

{task}


PREVIOUS CODE:

{draft}


CRITIC FEEDBACK:

{json.dumps(
    feedback,
    indent=2
)}


Produce the improved code.
"""
    }
]
    return call_llm(messages=messages) 
````

````python
def reflection_loop(task, max_iterations=3):
 

    # ---------------------------------------------------------
    # 1. Generate the initial solution
    # ---------------------------------------------------------
    draft = generate(task)

    print("=" * 70)
    print("INITIAL SOLUTION")
    print("=" * 70)
    print(draft)

    # ---------------------------------------------------------
    # 2. Reflection loop
    # ---------------------------------------------------------
    for iteration in range(1, max_iterations + 1):

        print("\n" + "=" * 70)
        print(f"REFLECTION ITERATION {iteration}")
        print("=" * 70)

        # -----------------------------------------------------
        # Critic analyzes the current solution
        # -----------------------------------------------------
        feedback = critic(
            task=task,
            draft=draft
        )

        print("\n--- CRITIC FEEDBACK ---")
        print(feedback)

        # -----------------------------------------------------
        # Improve the solution using the feedback
        # -----------------------------------------------------
        improved_draft = improve_(
            draft=draft,
            feedback=feedback,
            task=task
        )

        print("\n--- IMPROVED SOLUTION ---")
        print(improved_draft)

        # -----------------------------------------------------
        # Update the current solution
        # -----------------------------------------------------
        draft = improved_draft

    # ---------------------------------------------------------
    # 3. Return final solution
    # ---------------------------------------------------------
    return draft

### Usage

task = """
Write a Python function that reads a JSON file and returns
all users whose age is greater than 18.

The function should:
- handle a missing file
- handle invalid JSON
- use type hints
- be easy to test
"""

final_code = reflection_loop(
    task=task,
    max_iterations=3
)

print("\n" + "=" * 70)
print("FINAL RESULT")
print("=" * 70)
print(final_code)
````
